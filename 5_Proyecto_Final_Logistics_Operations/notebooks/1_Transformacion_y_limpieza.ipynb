# <span style="color:orange">Transformación y limpieza de datos<span>

```python
import pandas as pd

pd.set_option('display.max_columns', None)
```

```python
# Algunas funciones que utilizaremos en este archivo
def sep():
    print('---------------------------------------')

def revisar_duplicados(df, col):
    sep()
    dup = df[col].duplicated().sum()
    print(f'Los valores duplicados en la columna {col} es de {dup}')
```

## <span style="color:orange">---------  Archivo *df_loads_trips.csv*  ---------<span>

```python
# Carga de archivos en bruto
df_loads = pd.read_csv("../datos_en_bruto/loads.csv", sep = ',')
df_trips = pd.read_csv("../datos_en_bruto/trips.csv", sep = ',')
```

```python
print(f'El número de filas que tenemos en df_loads es {df_loads.shape[0]}, y el número de columnas es {df_loads.shape[1]}')
sep()
print (f'El número de filas que tenemos en df_trips es {df_trips.shape[0]}, y el número de columnas es {df_trips.shape[1]}')

# Nombre de las columnas de ambos archivos
sep()
print('df_loads:\n', df_loads.columns)
sep()
print('df_trips:\n', df_trips.columns)
```

```python
# df_trips['load_id'].head()
# df_trips['load_id'].tail()
# df_loads['load_id'].head()
df_loads['load_id'].tail()
```

<span style="color:orange">En una primera revisión, podemos comprobar que en el archivo trips, el **trip_id** y el **load_id** se corresponden uno a uno, así que vamos a hacer un merge de los dos archivos y trabajar solo con uno pero con más datos<span>

```python
df = df_loads.merge(df_trips, on='load_id', how='inner')
df.shape
print (f'El número de filas que tenemos es {df.shape[0]}, y el número de columnas es {df.shape[1]}')
```

```python
df.describe().round(2).T
```

Aquí podemos echar un vistazo a las métricas más importantes

<span style="color:orange">Revisamos los tipos de datos y echamos un primer vistazo a los nulos<span>

```python
df.info()
```

<span style="color:orange">**Revisión de:** <br>
    - nombres de las columnas <br>
    - datos únicos en las cada columna <br>
    - duplicados en load_id y trip_id
<span>

```python
print(df.columns)

for col in df.columns:
    sep()
    print(f'Columna: *** {col} ***:')
    print(df[col].unique())
    
revisar_duplicados(df, 'load_id')
revisar_duplicados(df, 'trip_id')
```

#### <span style="color:orange">**Limpieza de columnas** <br>
<span style="color:orange">Dentro de este archivo, hay columnas que no sirven para el análisis, y para que no ocupen espacio y no perjudique al rendimiento, las eliminamos. <br>Otras en cambio, vamos a añadirlas, como total_revenue, para utilizarla en el análisis más adelante
<span>

```python
df.drop(columns=['load_id', 'trip_id', 'load_status','trip_status'], inplace=True)
df['total_revenue'] = df['revenue'] - df['accessorial_charges'] - df['fuel_surcharge']
df.sample(5)
```

#### <span style="color:orange">**Limpieza de duplicados**
<span style="color:orange">no encontramos duplicados

```python
df.duplicated().sum()
```

#### <span style="color:orange">**Limpieza de nulos**
<span style="color:orange">Guardamos el archivo **df.csv** <br>
De momento no tocamos los nulos, pero los dejamos identificados

```python
df.isna().sum()
```

```python
df.to_csv('../datos_transformados/df_loads_trips.csv')
```

## <span style="color:orange">---------  Archivo *df_empleados.csv*  ---------
<span style="color:orange">Creamos este archivo para dar respuesta a los tres primeros puntos que nos pide el cliente: <br>
<span style="color:orange">- *1. Los 10 empleados que más viajes han realizado* <br>
<span style="color:orange">- *2. Los 10 empleados que más beneficios han reportado a la empresa* <br>
<span style="color:orange">- *3. Los 10 empleados que más tiempo han estado en la carretera*<span>

```python
df_empleados = df.copy(deep=True) # deep=True para que se copie y sea un archivo independiente
```

<span style="color:orange">Eliminamos las columnas que no vamos a necesitar

```python
df_empleados.drop(columns=['customer_id', 'route_id', 'load_date', 
                           'load_type', 'weight_lbs', 
                           'weight_lbs', 'pieces', 'booking_type',
                           'truck_id', 'trailer_id', 'fuel_gallons_used',
                           'average_mpg', ], inplace=True)
```

```python
df_empleados.columns
```

<span style="color:orange">Para este archivo (df_empleados), nos quedamos solo con los viajes realizados en el año 2024 >> nos quedamos solo con el año y descartamos mes y día

```python
df_empleados["dispatch_date"] = pd.to_datetime(df_empleados["dispatch_date"])
df_empleados['dispatch_date_year'] = df_empleados['dispatch_date'].dt.year
df_empleados = df_empleados[df_empleados['dispatch_date_year'] == 2024]
df_empleados.drop(columns=['dispatch_date'], inplace=True)
df_empleados.sample(5)
```

<span style="color:orange">Revisamos la columna driver_id, por si hay nulos, eliminarlos, y guardamos el archivo para el futuro análisis

```python
df_empleados['driver_id'].isna().sum()
df_empleados.dropna(subset=["driver_id"], inplace=True)
df_empleados.to_csv('../datos_transformados/df_empleados.csv')
```
